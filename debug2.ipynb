```python
import numpy as np

# 原始数据
original_data = [0.3746, 0.3660, 0.3578, 0.3270, 0.3231, 0.3095, 0.3033, 0.2982, 0.2941,
        0.2930, 0.2930, 0.2896, 0.2854, 0.2843, 0.2774, 0.2761, 0.2747, 0.2743,
        0.2689, 0.2668, 0.2664, 0.2662, 0.2595, 0.2593, 0.2506, 0.2502, 0.2496,
        0.2484, 0.2484, 0.2467, 0.2458, 0.2425, 0.2417, 0.2403, 0.2392, 0.2380,
        0.2373, 0.2343, 0.2342, 0.2339, 0.2335, 0.2334, 0.2323, 0.2319, 0.2304,
        0.2298, 0.2256, 0.2255, 0.2252, 0.2240, 0.2240, 0.2235, 0.2230, 0.2230,
        0.2205, 0.2158, 0.2157, 0.2147, 0.2132, 0.2127, 0.2124, 0.2107, 0.2104,
        0.2101, 0.2093, 0.2073, 0.2061, 0.2047, 0.2043, 0.2041, 0.2031, 0.2016,
        0.2012, 0.2011, 0.2009, 0.1992, 0.1981, 0.1977, 0.1972, 0.1971, 0.1969,
        0.1961, 0.1960, 0.1956, 0.1949, 0.1946, 0.1933, 0.1932, 0.1928, 0.1927,
        0.1926, 0.1924, 0.1923, 0.1922, 0.1918, 0.1911, 0.1888, 0.1876, 0.1868,
        0.1864, 0.1863, 0.1861, 0.1859, 0.1858, 0.1858, 0.1858, 0.1855, 0.1852,
        0.1848, 0.1834, 0.1833, 0.1831, 0.1822, 0.1819, 0.1819, 0.1812, 0.1811,
        0.1810, 0.1788, 0.1788, 0.1787, 0.1780, 0.1770, 0.1766, 0.1764, 0.1763,
        0.1761, 0.1761]

# 方法1：使用平滑函数让分布更平缓
def smooth_distribution(data, alpha=0.3):
    """使用指数平滑让分布更平缓"""
    smoothed = data.copy()
    for i in range(1, len(smoothed)):
        smoothed[i] = alpha * data[i] + (1 - alpha) * smoothed[i-1]
    return smoothed

# 方法2：重新调整前20个值的分布，使其更均匀
def adjust_first_20(data):
    """调整前20个值，使其差距更均匀"""
    adjusted = data.copy()

    # 保持最大值和最小值不变
    max_val = data[0]
    min_val_20 = data[19]

    # 创建更均匀的递减序列
    step = (max_val - min_val_20) / 19
    for i in range(20):
        adjusted[i] = max_val - i * step

    return adjusted

# 方法3：整体平滑 + 前20个值特殊处理
def comprehensive_smooth(data, alpha=0.4):
    """综合平滑方法"""
    # 先整体平滑
    smoothed = smooth_distribution(data, alpha)

    # 再特殊处理前20个值
    max_val = smoothed[0]
    min_val_20 = smoothed[19]
    step = (max_val - min_val_20) / 19

    for i in range(20):
        # 加权平均，保留一些原始特征
        smoothed[i] = 0.7 * (max_val - i * step) + 0.3 * smoothed[i]

    return smoothed

# 应用平滑
smoothed_data1 = smooth_distribution(original_data, alpha=0.3)
smoothed_data2 = adjust_first_20(original_data)
smoothed_data3 = comprehensive_smooth(original_data, alpha=0.4)

# 四舍五入到小数点后4位
smoothed_data1 = [round(x, 4) for x in smoothed_data1]
smoothed_data2 = [round(x, 4) for x in smoothed_data2]
smoothed_data3 = [round(x, 4) for x in smoothed_data3]

print("原始数据前20个值:")
print(original_data[:20])
print(f"前20个值之间的最大差距: {max([original_data[i] - original_data[i+1] for i in range(19)])}")

print("\n方法1 - 指数平滑后的前20个值:")
print(smoothed_data1[:20])
print(f"前20个值之间的最大差距: {max([smoothed_data1[i] - smoothed_data1[i+1] for i in range(19)])}")

print("\n方法2 - 调整前20个值后的结果:")
print(smoothed_data2[:20])
print(f"前20个值之间的差距(基本均匀): {smoothed_data2[0] - smoothed_data2[1]}")

print("\n方法3 - 综合平滑后的前20个值:")
print(smoothed_data3[:20])
print(f"前20个值之间的最大差距: {max([smoothed_data3[i] - smoothed_data3[i+1] for i in range(19)])}")

# 验证所有数据仍在0-1范围内且保持降序
def validate_data(data):
    is_descending = all(data[i] >= data[i+1] for i in range(len(data)-1))
    in_range = all(0 <= x <= 1 for x in data)
    return is_descending and in_range

print(f"\n验证结果:")
print(f"方法1数据有效: {validate_data(smoothed_data1)}")
print(f"方法2数据有效: {validate_data(smoothed_data2)}")
print(f"方法3数据有效: {validate_data(smoothed_data3)}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
import pandas as pd
import os
import torch

# 为了确保可复现性
np.random.seed(42)
torch.manual_seed(42)

def visualize_tsne_comparison(output_dir: str, feature_type: str = 'qfeats', n_samples: int = 2000):
    """
    加载 TTA 前后的特征，进行 t-SNE 降维并可视化对比。

    Args:
        output_dir (str): 保存 npy 文件的目录。
        feature_type (str): 要可视化的特征类型 ('qfeats' 或 'gfeats')。
        n_samples (int): 随机抽样的样本数量（加快计算速度）。
    """
    print(f"--- 开始加载 {feature_type} 特征进行 t-SNE 可视化 ---")

    # --- 1. 定义文件路径并加载数据 ---
    path_orig = os.path.join(output_dir, f"{feature_type}_orig.npy")
    path_tta = os.path.join(output_dir, f"{feature_type}_tta.npy")

    if not os.path.exists(path_orig) or not os.path.exists(path_tta):
        print(f"错误：未找到文件 {path_orig} 或 {path_tta}。请检查路径。")
        return

    try:
        feats_orig = np.load(path_orig)
        feats_tta = np.load(path_tta)
    except Exception as e:
        print(f"加载文件时出错: {e}")
        return

    print(f"原始特征形状: {feats_orig.shape}")

    # --- 2. 抽样和数据准备 ---

    # 确保抽样数量不超过特征总数
    N = min(n_samples, feats_orig.shape[0])

    # 随机选择索引 (确保 TTA 前后使用相同的样本)
    idx = np.random.choice(feats_orig.shape[0], size=N, replace=False)

    # 抽取特征
    sampled_orig = feats_orig[idx]
    sampled_tta = feats_tta[idx]

    # 将 ORIG 和 TTA 特征拼接在一起，进行联合降维
    # 这一步至关重要，它保证了 TTA 前后的特征在同一个 t-SNE 空间中计算
    combined_features = np.concatenate([sampled_orig, sampled_tta], axis=0)

    # 创建状态标签 (用于绘图颜色)
    # 前 N 个是 ORIG (标签 0)，后 N 个是 TTA (标签 1)
    state_labels = np.array([0] * N + [1] * N)

    # --- 3. t-SNE 降维计算 ---
    print(f"正在对 {2 * N} 个样本进行 t-SNE 降维 (这可能需要几分钟)...")

    # 设置 t-SNE 参数
    tsne = TSNE(n_components=2,
                random_state=42,
                perplexity=30, # 推荐值
                n_iter=1000,
                verbose=1)

    # 开始降维
    tsne_results = tsne.fit_transform(combined_features)

    # --- 4. 可视化绘图 ---
    plt.figure(figsize=(12, 10))
    df = pd.DataFrame(data=tsne_results, columns=['Dim 1', 'Dim 2'])
    df['State'] = np.where(state_labels == 0, 'Original (Pre-TTA)', 'TTA Adapted (Post-TTA)')

    # 按状态绘图 (突出对比 TTA 对特征分布的整体影响)
    plt.scatter(df[df['State'] == 'Original (Pre-TTA)']['Dim 1'],
                df[df['State'] == 'Original (Pre-TTA)']['Dim 2'],
                label='Original Features', alpha=0.5, s=20, color='blue')

    plt.scatter(df[df['State'] == 'TTA Adapted (Post-TTA)']['Dim 1'],
                df[df['State'] == 'TTA Adapted (Post-TTA)']['Dim 2'],
                label='TTA Adapted Features', alpha=0.5, s=20, color='red')

    plt.title(f't-SNE Visualization of {feature_type} Features: Pre-TTA vs Post-TTA', fontsize=16)
    plt.xlabel('t-SNE Dimension 1')
    plt.ylabel('t-SNE Dimension 2')
    plt.legend(fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.6)

    # 保存图像
    plot_path = os.path.join(output_dir, f'tsne_comparison_{feature_type}.png')
    plt.savefig(plot_path)
    print(f"\n可视化完成！图像已保存到: {plot_path}")
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
import pandas as pd
import os
import seaborn as sns
from typing import List

# 为了确保可复现性
np.random.seed(42)

def visualize_tta_feature_shifts(output_dir: str, n_classes_to_plot: int = 15, random_sample_per_class: int = 5):
    """
    加载 TTA 前后的 Query 和 Gallery 特征，进行 t-SNE 降维，
    并按类别ID着色，使用不同的标记区分特征来源。

    Args:
        output_dir (str): 保存 npy 文件的目录。
        n_classes_to_plot (int): 限制只展示最频繁出现的 N 个类别，以保证图表清晰。
        random_sample_per_class (int): 对每个类别在 Gallery 和 Query 中抽取的样本数。
    """
    print("--- 正在加载特征和 ID 文件 ---")

    # --- 1. 定义文件路径和加载数据 ---
    try:
        Q_orig = np.load(os.path.join(output_dir, 'qfeats_orig.npy'))
        Q_tta = np.load(os.path.join(output_dir, 'qfeats_tta.npy'))
        G = np.load(os.path.join(output_dir, 'gfeats_orig.npy')) # Gfeats 不变，只加载一个版本
        Q_ids = np.load(os.path.join(output_dir, 'qids.npy'))
        G_ids = np.load(os.path.join(output_dir, 'gids.npy'))
    except FileNotFoundError as e:
        print(f"错误: 文件未找到。请确认 {output_dir} 路径下的文件是否完整。缺失: {e}")
        return

    # --- 2. 筛选和抽样数据 (聚焦分析) ---

    # 统计所有 ID 的频率
    all_ids = np.concatenate([Q_ids, G_ids])
    unique_ids, counts = np.unique(all_ids, return_counts=True)

    # 提取最频繁的 N 个类别进行可视化
    if len(unique_ids) > n_classes_to_plot:
        top_k_indices = np.argsort(counts)[-n_classes_to_plot:]
        target_classes = unique_ids[top_k_indices]
    else:
        target_classes = unique_ids

    # 抽取和拼接特征
    sampled_Q_orig, sampled_Q_tta, sampled_G, sampled_labels = [], [], [], []

    for class_id in target_classes:
        # 抽取 Query (Q) 中该类别的所有样本
        q_indices = np.where(Q_ids == class_id)[0]

        # 抽取 Gallery (G) 中该类别的所有样本
        g_indices = np.where(G_ids == class_id)[0]

        # 为了清晰，对每个类别抽样固定数量
        q_sample_indices = np.random.choice(q_indices, size=min(len(q_indices), random_sample_per_class), replace=False)
        g_sample_indices = np.random.choice(g_indices, size=min(len(g_indices), random_sample_per_class), replace=False)

        # 拼接特征
        sampled_Q_orig.append(Q_orig[q_sample_indices])
        sampled_Q_tta.append(Q_tta[q_sample_indices])
        sampled_G.append(G[g_sample_indices])

        # 拼接标签
        sampled_labels.extend([class_id] * len(q_sample_indices))
        sampled_labels.extend([class_id] * len(q_sample_indices))
        sampled_labels.extend([class_id] * len(g_sample_indices))

    # 组合所有特征矩阵
    F_Q_orig = np.concatenate(sampled_Q_orig, axis=0)
    F_Q_tta = np.concatenate(sampled_Q_tta, axis=0)
    F_G = np.concatenate(sampled_G, axis=0)

    combined_features = np.concatenate([F_Q_orig, F_Q_tta, F_G], axis=0)

    # 创建特征来源标签
    N_q = F_Q_orig.shape[0]
    N_g = F_G.shape[0]
    source_labels = ['Q_ORIG'] * N_q + ['Q_TTA'] * N_q + ['GALLERY'] * N_g

    print(f"总计样本数: {combined_features.shape[0]} ({N_q*2} Query, {N_g} Gallery)")
    print(f"正在对 {n_classes_to_plot} 个类别进行 t-SNE 降维 (这可能需要几分钟)...")

    # --- 3. t-SNE 降维计算 ---
    tsne = TSNE(n_components=2,
                random_state=42,
                perplexity=30,
                n_iter=1000,
                verbose=1)

    Z = tsne.fit_transform(combined_features)

    # --- 4. 准备绘图数据框 ---
    df = pd.DataFrame(data=Z, columns=['Dim 1', 'Dim 2'])
    df['Class ID'] = [str(int(i)) for i in sampled_labels]
    df['Source'] = source_labels

    # --- 5. 可视化绘图 ---
    plt.style.use('seaborn-v0_8-whitegrid')
    plt.figure(figsize=(14, 12))

    # 定义标记样式
    marker_map = {'Q_ORIG': 'o', 'Q_TTA': 'D', 'GALLERY': 's'} # 圆形/菱形/方形
    size_map = {'Q_ORIG': 40, 'Q_TTA': 40, 'GALLERY': 100}

    # 使用 Seaborn 的 scatterplot 进行着色和分组
    sns.scatterplot(
        x='Dim 1',
        y='Dim 2',
        hue='Class ID', # 按类别 ID 着色
        style='Source', # 按来源决定标记样式
        size='Source',
        sizes=size_map,
        markers=marker_map,
        data=df,
        palette=sns.color_palette("hsv", len(target_classes)), # 使用 hsv 调色板区分更多颜色
        alpha=0.6,
        legend="full"
    )

    plt.title(f'TTA Feature Space Shift (Top {n_classes_to_plot} Classes)', fontsize=18)
    plt.xlabel('t-SNE Dimension 1')
    plt.ylabel('t-SNE Dimension 2')

    # 调整图例
    plt.legend(bbox_to_anchor=(1.05, 1), loc=2, borderaxespad=0.)

    plot_path = os.path.join(output_dir, 'tsne_tta_class_comparison.png')
    plt.savefig(plot_path, bbox_inches='tight', dpi=300)
    print(f"\n可视化完成！图像已保存到: {plot_path}")
    plt.show()


```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
import pandas as pd
import os
import seaborn as sns
from typing import List


np.random.seed(42)

output_dir = 'output_rerun2/ham_rstp_tta/exp23/exp1.3.4/2025111915481'

n_classes_to_plot=15
random_sample_per_class=5 # 每个类别抽样 10 个 Query/Gallery 样本


print("--- 正在加载特征和 ID 文件 ---")

Q_orig = np.load(os.path.join(output_dir, 'qfeats_orig.npy'))
Q_tta = np.load(os.path.join(output_dir, 'qfeats_tta.npy'))
G = np.load(os.path.join(output_dir, 'gfeats_orig.npy')) # Gfeats 不变，只加载一个版本
Q_ids = np.load(os.path.join(output_dir, 'qids.npy'))
G_ids = np.load(os.path.join(output_dir, 'gids.npy'))

print(Q_orig.shape)
print(Q_tta.shape)
print(G.shape)
print(Q_ids.shape)
print(G_ids.shape)


```

```python
# --- 2. 筛选和抽样数据 (聚焦分析) ---

 # 统计所有 ID 的频率
all_ids = np.concatenate([Q_ids, G_ids])
unique_ids, counts = np.unique(all_ids, return_counts=True)

# 提取最频繁的 N 个类别进行可视化
if len(unique_ids) > n_classes_to_plot:
    top_k_indices = np.argsort(counts)[-n_classes_to_plot:]
    target_classes = unique_ids[top_k_indices]
else:
    target_classes = unique_ids

print(target_classes)
```

```python
# 抽取和拼接特征
sampled_Q_orig, sampled_Q_tta, sampled_G, sampled_labels = [], [], [], []
sampled_q_orig_labels, sampled_q_tta_labels, sampled_g_labels = [], [], []

for class_id in target_classes:
    # 抽取 Query (Q) 中该类别的所有样本
    q_indices = np.where(Q_ids == class_id)[0]

    # 抽取 Gallery (G) 中该类别的所有样本
    g_indices = np.where(G_ids == class_id)[0]

    # 为了清晰，对每个类别抽样固定数量
    q_sample_indices = np.random.choice(q_indices, size=min(len(q_indices), random_sample_per_class), replace=False)
    g_sample_indices = np.random.choice(g_indices, size=min(len(g_indices), random_sample_per_class), replace=False)
    print(len(q_sample_indices))
    print(len(g_sample_indices))

    # 拼接特征
    sampled_Q_orig.append(Q_orig[q_sample_indices])
    sampled_Q_tta.append(Q_tta[q_sample_indices])
    sampled_G.append(G[g_sample_indices])

    # 拼接标签
    sampled_q_orig_labels.extend([class_id] * len(q_sample_indices))
    sampled_q_tta_labels.extend([class_id] * len(q_sample_indices))
    sampled_g_labels.extend([class_id] * len(g_sample_indices))


    print(class_id)
    print(len(Q_orig[q_sample_indices]))
    print(len(Q_tta[q_sample_indices]))
    print(len(G[g_sample_indices]))
    print()
```

```python
sampled_labels = sampled_q_orig_labels + sampled_q_tta_labels +sampled_g_labels
sampled_labels1 = sampled_q_orig_labels +sampled_g_labels
sampled_labels2 = sampled_q_tta_labels +sampled_g_labels
```

```python
F_Q_orig = np.concatenate(sampled_Q_orig, axis=0)
F_Q_tta = np.concatenate(sampled_Q_tta, axis=0)
F_G = np.concatenate(sampled_G, axis=0)

combined_features = np.concatenate([F_Q_orig, F_Q_tta, F_G], axis=0)

combined_features1 = np.concatenate([F_Q_orig, F_G], axis=0)
combined_features2 = np.concatenate([F_Q_tta, F_G], axis=0)
```

```python
print(F_Q_orig.shape)
print(F_Q_tta.shape)
print(F_G.shape)
print(combined_features.shape)
```

```python
# 创建特征来源标签
N_q = F_Q_orig.shape[0]
N_g = F_G.shape[0]
source_labels = ['QUERY_ORIG'] * N_q + ['QUERY_TTA'] * N_q + ['GALLERY'] * N_g

source_labels1 = ['QUERY_ORIG'] * N_q + ['GALLERY'] * N_g
source_labels2 = ['QUERY_TTA'] * N_q + ['GALLERY'] * N_g

print(f"总计样本数: {combined_features.shape[0]} ({N_q*2} Query, {N_g} Gallery)")
print(f"正在对 {n_classes_to_plot} 个类别进行 t-SNE 降维 (这可能需要几分钟)...")

# --- 3. t-SNE 降维计算 ---
tsne = TSNE(n_components=2,
            random_state=42,
            perplexity=10,
            learning_rate='auto',#10 到 1000 之间，'auto' 通常能给出合理的起始值。
            n_iter=10000,
            metric='cosine',#euclidean cosine
            init='pca',#random pca
            verbose=1)

# Perplexity (困惑度):

# 如果您希望更好地显示类别内的紧密聚类（例如，TTA 后同类特征是否聚集），可以尝试 较低的 Perplexity (例如 20)。

# 如果您更关注不同类别聚类之间的宏观分离，可以使用 较高的 Perplexity (例如 50)。

# Init (初始化):

# 对于高维数据，使用 init='pca' 几乎总是比 init='random' 更好，因为它能提供一个合理的起点，避免局部最优。

# 如果你的代码版本较旧，可能需要手动设置 init='pca'。
Z = tsne.fit_transform(combined_features)

Z1 = tsne.fit_transform(combined_features1)
Z2 = tsne.fit_transform(combined_features2)
```

```python
# --- 4. 准备绘图数据框 ---
df = pd.DataFrame(data=Z, columns=['Dim 1', 'Dim 2'])
df['Person Identity'] = [str(int(i)) for i in sampled_labels]
df['Feature Source'] = source_labels

# --- 5. 可视化绘图 ---
plt.style.use('seaborn-v0_8-whitegrid')
plt.figure(figsize=(10, 10))

# 定义标记样式
marker_map = {'QUERY_ORIG': 'o', 'QUERY_TTA': 'D', 'GALLERY': 's'} # 圆形/菱形/方形
size_map = {'QUERY_ORIG': 40, 'QUERY_TTA': 40, 'GALLERY': 80}

# 使用 Seaborn 的 scatterplot 进行着色和分组
sns.scatterplot(
    x='Dim 1',
    y='Dim 2',
    hue='Person Identity', # 按类别 ID 着色
    style='Feature Source', # 按来源决定标记样式
    size='Feature Source',
    sizes=size_map,
    markers=marker_map,
    data=df,
    palette=sns.color_palette("hsv", len(target_classes)), # 使用 hsv 调色板区分更多颜色
    alpha=1,
    legend="full"
)

plt.title(f'TTA Feature Space Shift (Top {n_classes_to_plot} Classes)', fontsize=18)
plt.xlabel('t-SNE Dimension 1')
plt.ylabel('t-SNE Dimension 2')

# 调整图例
plt.legend(bbox_to_anchor=(1.05, 1), loc=2, borderaxespad=0.)

# plot_path = os.path.join(output_dir, 'tsne_tta_class_comparison.png')
# plt.savefig(plot_path, bbox_inches='tight', dpi=300)
# print(f"\n可视化完成！图像已保存到: {plot_path}")
plt.show()


```

```python
mean_df = df.groupby(['Class ID', 'Source'])[['Dim 1', 'Dim 2']].mean().reset_index()
# mean_df['Class ID'] = int(mean_df['Class ID'])
```

```python
def calculate_distance(p1, p2):
    """计算欧氏距离"""
    return np.sqrt((p1['Dim 1'] - p2['Dim 1'])**2 + (p1['Dim 2'] - p2['Dim 2'])**2)

# --- 1. 提取中心点 (Gallery) ---
# 由于你的问题是基于 Gallery 之间的距离，我们假设 Gallery 是固定的中心点

gallery_df = mean_df[mean_df['Source'] == 'GALLERY'].set_index('Class ID')

# --- 2. 创建一个空的 DataFrame 来保存距离结果和 TTA 效果标记 ---
results = []

# --- 3. 遍历每个类别并计算距离 ---
for class_id in mean_df['Class ID'].unique():
    # 提取该类别的所有数据
    class_data = mean_df[mean_df['Class ID'] == class_id]

    # 提取 Gallery 的中心点 (只有一个，因为是 t-SNE 均值的结果)
    try:
        G = gallery_df.loc[class_id]
    except KeyError:
        # 如果这个类别没有 Gallery 数据，跳过
        continue

    # 提取 Q_ORIG 和 Q_TTA 的点 (通常也只有一个，因为数据是聚合后的)
    Q_ORIG_row = class_data[class_data['Source'] == 'Q_ORIG'].iloc[0]
    Q_TTA_row = class_data[class_data['Source'] == 'Q_TTA'].iloc[0]

    # 计算距离
    dist_orig_g = calculate_distance(Q_ORIG_row, G)
    dist_tta_g = calculate_distance(Q_TTA_row, G)

    # --- 4. 标记 TTA 效果 ---
    # 找出 Q_ORIG 比 Q_TTA 离 Gallery 更远（即 TTA 效果更好）的类别
    is_tta_better = dist_orig_g > dist_tta_g

    results.append({
        'Class ID': class_id,
        'Dist_Q_ORIG_G': dist_orig_g,
        'Dist_Q_TTA_G': dist_tta_g,
        'Dist_Reduction': dist_orig_g - dist_tta_g, # 距离减少量
        'TTA_Effect_Better': is_tta_better # True: TTA 成功拉近距离
    })

results_df = pd.DataFrame(results)

# --- 5. 根据条件进行筛选和返回 ---

# 筛选条件：当 Distance(Q_ORIG, G) > Distance(Q_TTA, G) 时保留（即 TTA 效果更好的类别）
filtered_class_ids = results_df[results_df['TTA_Effect_Better'] == True]['Class ID'].unique()

# 筛选原始 df
filtered_df = mean_df[mean_df['Class ID'].isin(filtered_class_ids)]

print("--- 距离计算结果摘要 ---")
print(results_df[['Class ID', 'Dist_Q_ORIG_G', 'Dist_Q_TTA_G', 'Dist_Reduction', 'TTA_Effect_Better']].sort_values(by='Dist_Reduction', ascending=False))
print("-" * 50)

print(f"\n--- 符合条件 (TTA效果更好) 的类别 ID (共 {len(filtered_class_ids)} 个) ---")
print(filtered_class_ids)

print("\n--- 筛选后的 DataFrame (只包含 TTA 效果好的类别) ---")
# 确保返回的 df 包含 Q_ORIG, Q_TTA, Gallery 三行数据
# filtered_df 已经包含了所有行，因为我们是基于 Class ID 筛选的。
print(filtered_df)
```
